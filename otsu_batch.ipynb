```python
def otsu(img8):
    scores = np.full(255, np.inf)

    for T in range(1, 256):
        class1 = img8[img8 < T]
        class2 = img8[img8 >= T]

        if class1.size == 0 or class2.size == 0:
            continue

        scores[T - 1] = (
            class1.size * np.var(class1)
            + class2.size * np.var(class2)
        )

    T_opt = np.argmin(scores) + 1

    return T_opt
```

**[Task 2.2.]()** Implement a re-usable function to compute the Dice coefficient:

```python
def compute_dice(bin1, bin2):
    assert bin1.dtype == bool
    assert bin2.dtype == bool
    
    dice = 2 * (bin1 & bin2).sum() / (bin1.sum() + bin2.sum())
    
    return dice
```

**[Task 2.4.]()** Perform batch processing:

```python
dice_scores = []

for i in [28, 29, 33, 44, 46, 49]:

    imgf = plt.imread(f"data/NIH3T3/im/dna-{i}.png")

    img8 = (imgf * 255).round().astype(np.uint8)

    T_opt = otsu(img8)
    segmentation = img8 >= T_opt

    h = plt.imread(f"data/NIH3T3/gt/{i}.png")
    h_bool = h > 0

    dice = compute_dice(segmentation, h_bool)

    dice_scores.append(dice)
    print(f"Image {i}: Dice = {dice:.4f}")

mean_dice = np.mean(dice_scores)

print(f"Mean Dice coefficient: {mean_dice:.4f}")
```
